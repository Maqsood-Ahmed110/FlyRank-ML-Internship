## 2. The question: decision, action, cost of a wrong call

**Decision improved:** which pages enter a content team's limited weekly review
capacity — instead of reviewing by gut feel or alphabetically, they get a ranked,
reason-coded queue.

**Action taken:** a human reviewer opens the top-N pages, checks the reason code
(e.g. stale_visible_page, declining_with_demand, low_ctr_visible_page), and decides
whether to refresh, expand, or leave the page alone. The model prioritizes attention;
it doesn't act on its own.

**Cost of a wrong call:** a false positive wastes a reviewer's limited time on a page
that wasn't really worth fixing. A false negative means a genuinely declining page
goes unnoticed. Because review capacity is limited, precision at the top of the
ranking matters more than overall accuracy — which is why this project should be
judged on Precision@K, not accuracy.
