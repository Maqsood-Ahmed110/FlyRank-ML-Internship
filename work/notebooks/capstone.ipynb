## 3. Methodology

## 3. Methodology

### Model

I trained a **Random Forest classifier** using page-level content and performance features.

### Features

The model uses available page-level signals that describe content and observed performance. Features were selected to represent information that could reasonably support a content-review prioritization decision.

### Label

The classification target represents whether an observation belongs to the review-priority outcome defined during the project analysis.

### Baseline

A naive random train/test split was used as a comparison point. This provides a useful reference but may produce an optimistic estimate when related observations from the same client appear in both training and evaluation data.

### Validation Design

The primary evaluation uses a **client-grouped validation split**. Client groups are kept separated between training and evaluation so that the model is tested on client groups it did not see during training.

This validation design is more appropriate for assessing whether the approach can generalize beyond the groups represented in training.

### Leakage Checks

The analysis avoids using evaluation outcomes or information derived from the target as model inputs. Client grouping is applied during validation rather than being treated as an ordinary predictive feature.

The grouped validation result is treated as the primary evidence because it better reflects the intended unseen-client scenario.

## 4. Results (vs baseline)

The model was evaluated using both a naive random split and the more realistic client-grouped validation design.

| Evaluation                |  Precision |         F1 |
| ------------------------- | ---------: | ---------: |
| Naive random split        |     0.7446 |          — |

The random split produced a higher precision of **0.7446**. However, the client-grouped evaluation produced a lower precision of **0.5876** and an F1 score of **0.2032**.

The grouped result is treated as the more informative estimate because it evaluates the model on unseen client groups. The difference between the two evaluations demonstrates that a conventional random split can provide an overly optimistic view of model performance.

The result supports using the model as a **directional prioritization signal**, rather than claiming that it provides a definitive or autonomous content decision.


## 5. Limitations

*What this work cannot claim.*

This analysis has several limitations.

1. **Limited evaluation scope:** The model was evaluated on the available project data and should not automatically be assumed to generalize to all future pages, clients, or content environments.

2. **Moderate precision:** The grouped validation precision of **0.5876** means that a substantial portion of prioritized observations would not correspond to the target outcome. The model should therefore support human review rather than replace it.

3. **Low F1:** The grouped validation F1 score of **0.2032** indicates that the model does not provide a complete solution to the classification problem.

4. **Validation sensitivity:** The difference between the random-split precision (**0.7446**) and grouped precision (**0.5876**) demonstrates that evaluation design materially affects the measured result.

5. **No causal claim:** The model identifies associations in the available data. It does not demonstrate that changing a page based on a model recommendation will cause improved performance.

6. **Decision-support only:** The model should not independently publish, modify, remove, or make final decisions about content.

Future work should test the approach on additional future observations and unseen client groups and measure whether the resulting prioritization actually improves review efficiency or content outcomes.
